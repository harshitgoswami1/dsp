```python
import base64
import codecs
from urllib.parse import quote, unquote

def obfuscate(text, method):
    if method == "base64":
        return base64.b64encode(text.encode()).decode()
    elif method == "hex":
        return text.encode().hex()
    elif method == "rot13":
        return codecs.encode(text,"rot13")
    elif method == "url":
        return quote(text)

def deobfuscate(text, method):
    if method == "base64":
        return base64.b64decode(text).decode()
    elif method == "hex":
        return bytes.fromhex(text).decode()
    elif method == "rot13":
        return codecs.decode(text,"rot13")
    elif method == "url":
        return unquote(text)

encoded = obfuscate("hello world", "rot13")
print("Encoded:", encoded)

decoded = deobfuscate(encoded, "rot13")
print("Decoded:", decoded)
```

```python

import hashlib
from cryptography.fernet import Fernet

# Input
message = "Hi there"
changed = "Hi there please send me your details"

# 1. Confidentiality
key = Fernet.generate_key()
cipher = Fernet(key)

message = message.encode()

encrypted = cipher.encrypt(message)
decrypted = cipher.decrypt(encrypted).decode()

print("\nCONFIDENTIALITY")
print("Encrypted:", encrypted)
print("Decrypted:", decrypted)

# 2. Integrity
hash1 = hashlib.new("SHA256",message).hexdigest()
hash2 = hashlib.new("SHA256",changed.encode()).hexdigest()

print("\nINTEGRITY")
print("Original hash:", hash1)
print("Changed hash:", hash2)
print("Data unchanged:", hash1 == hash2)


```

```python

from sklearn.ensemble import RandomForestClassifier

# Training URLs
urls = [
    "https://google.com",
    "https://python.org",
    "https://wikipedia.org",
    "http://secure-login-example.com",
    "http://verify-account-example.com",
    "http://bank-update-example.com"
]

# Labels
y = ["legitimate", "legitimate", "legitimate",
     "phishing", "phishing", "phishing"]

# Extract features: URL length and hyphens
X = [[len(url), url.count("-"),url.count(".")] for url in urls]

# Train model
model = RandomForestClassifier(random_state=42)
model.fit(X, y)

# Test a new URL
url = input("Enter URL: ")
features = [[len(url), url.count("-"),url.count(".")]]

# Predict
print("Prediction:", model.predict(features)[0])
```

```python

from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import rsa, padding

# Generate keys
private = rsa.generate_private_key(public_exponent=65537, key_size=2048)
public = private.public_key()

# Message
msg = b"Hello World"

# Padding
pad = padding.PSS(
    mgf=padding.MGF1(hashes.SHA256()),
    salt_length=padding.PSS.MAX_LENGTH
)

pad1 = padding.PKCS1v15()

# Sign
signature = private.sign(msg, pad, hashes.SHA256())
print("Signature created")

# Verify
try:
    public.verify(signature, msg, pad, hashes.SHA256())
    print("Valid Signature")
except:
    print("Invalid Signature")
```

```python
import hashlib

users = []

def register(username, password):
    password = hashlib.new("sha256", password.encode()).hexdigest()
    users.append({"username": username, "password": password})
    print("Registration successful")


def login(username, password):
    password = hashlib.new("sha256", password.encode()).hexdigest()

    for user in users:
        if user["username"] == username and user["password"] == password:
            print("Login successful")
            return

    print("Invalid username or password")


register("student", "hello123")

login("student", "hello123")
login("student", "wrong123")
```

```python

from collections import Counter
import random

data = [
    {"age": 21, "city": "Delhi", "disease": "Flu"},
    {"age": 21, "city": "Delhi", "disease": "Cold"},
    {"age": 25, "city": "Mumbai", "disease": "Flu"},
    {"age": 25, "city": "Mumbai", "disease": "Cold"}
]

# Generalize age
for row in data:
    row["age"] = "20-29"

# Group rows by quasi-identifiers
groups = {}

for row in data:
    key = (row["age"], row["city"])
    groups.setdefault(key, []).append(row)

# 1. K-anonymity
k = 2
k_anonymous = all(len(g) >= k for g in groups.values())
print("K-anonymous:", k_anonymous)

# 2. L-diversity
l = 2
l_diverse = all(
    len(set(row["disease"] for row in g)) >= l
    for g in groups.values()
)
print("L-diverse:", l_diverse)

# 3. T-closeness
t = 0.5
all_diseases = Counter(row["disease"] for row in data)
total = len(data)

t_close = True

for g in groups.values():
    local = Counter(row["disease"] for row in g)

    distance = 0.5 * sum(
        abs(local[x] / len(g) - all_diseases[x] / total)
        for x in all_diseases
    )

    if distance > t:
        t_close = False

print("T-close:", t_close)

# 4. Differential Privacy (simple demonstration)
epsilon = 1.0
noisy_count = total + random.uniform(-1, 1)
print("Noisy count:", noisy_count)
```
